```python
import os
import json

DATA_DIR = "/content/squad_data"

print("当前数据目录：")
print(os.listdir(DATA_DIR))

for filename in [
    "train-v2.0.json",
    "dev-v2.0.json",
    "evaluate-v2.0.py"
]:
    path = os.path.join(DATA_DIR, filename)
    if os.path.exists(path):
        print(f"✅ {filename}: {os.path.getsize(path) / 1024 / 1024:.2f} MB")
    else:
        print(f"❌ {filename}: 文件不存在")
```

```python
with open(f"{DATA_DIR}/train-v2.0.json", "r", encoding="utf-8") as f:
    train_raw = json.load(f)

with open(f"{DATA_DIR}/dev-v2.0.json", "r", encoding="utf-8") as f:
    dev_raw = json.load(f)

print("Train articles:", len(train_raw["data"]))
print("Dev articles:", len(dev_raw["data"]))

print("Train version:", train_raw.get("version"))
print("Dev version:", dev_raw.get("version"))
```
